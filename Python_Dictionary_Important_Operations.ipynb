## 1. Creating dictionaries

Use `{key: value}` or `dict()`. Repeated keys in a literal keep the last value.

```python
student = {"name": "Aman", "age": 20, "course": "Python"}
empty = {}
from_pairs = dict([("name", "Riya"), ("age", 21)])
duplicate_key = {"score": 80, "score": 95}

print(student, type(student))
print(empty, from_pairs)
print(duplicate_key)
```

## 2. Accessing values: `[]` and `get()`

`d[key]` raises `KeyError` if the key is absent. `d.get(key)` returns `None` when absent; `d.get(key, default)` returns your default.

```python
student = {"name": "Aman", "age": 20}
print(student["name"])
print(student.get("age"))
print(student.get("marks"))
print(student.get("marks", 0))

if "course" in student:
    print(student["course"])
else:
    print("Course not available")
```

## 3. Adding and updating

Assigning to a new key adds it; assigning to an existing key changes its value.

```python
student = {"name": "Aman", "age": 20}
student["course"] = "Python"  # add
student["age"] = 21           # update
student["marks"] = 95
print(student)
```

## 4. Removing entries

- `pop(key)`: remove and return a value; missing keys raise `KeyError` unless a default is supplied.
- `popitem()`: remove and return the last inserted pair.
- `del d[key]`: delete a key.
- `clear()`: remove all entries.

```python
student = {"name": "Aman", "age": 20, "course": "Python"}
age = student.pop("age")
print("popped:", age, student)
print(student.pop("marks", 0))  # safe default
print(student.popitem())
del student["name"]
print(student)
student.clear()
print(student)
```

## 5. `keys()`, `values()`, and `items()`

These return dynamic view objects. Use `list(...)` if you need a list snapshot.

```python
student = {"name": "Aman", "age": 20, "course": "Python"}
print(student.keys())
print(student.values())
print(student.items())
print(list(student.keys()))
print(list(student.values()))
print(list(student.items()))
```

## 6. Iteration and membership

Iterating directly over a dictionary yields keys. `key in d` checks keys, not values.

```python
marks = {"Aman": 90, "Riya": 95, "Kabir": 78}

for name in marks:
    print(name)
for score in marks.values():
    print(score)
for name, score in marks.items():
    print(name, score)

print("Riya" in marks)
print(95 in marks)           # checks keys: False
print(95 in marks.values())  # checks values: True
```

## 7. `update()` and `setdefault()`

`update()` adds or overwrites entries. `setdefault(key, default)` returns the existing value or inserts the default if the key is missing.

```python
student = {"name": "Aman", "age": 20}
student.update({"age": 21, "course": "Python"})
print(student)

print(student.setdefault("age", 0))       # existing value
print(student.setdefault("city", "Delhi"))  # inserts city
print(student)
```

## 8. Copying and merging

`copy()` makes a shallow copy. In Python 3.9+, `a | b` creates a merged dictionary, with the right-hand value winning for duplicate keys. `update()` modifies the left-hand dictionary.

```python
original = {"name": "Aman", "marks": 90}
copied = original.copy()
copied["marks"] = 100
print("original:", original)
print("copy:", copied)

a = {"x": 1, "y": 2}
b = {"y": 99, "z": 3}
print("merged:", a | b)
a.update(b)
print("a after update:", a)
```

## 9. Dictionary comprehensions

Pattern: `{key_expression: value_expression for item in iterable if condition}`.

```python
numbers = [1, 2, 3, 4, 5]
squares = {n: n * n for n in numbers}
even_squares = {n: n * n for n in numbers if n % 2 == 0}
names = ["aman", "riya", "kabir"]
lengths = {name: len(name) for name in names}

print(squares)
print(even_squares)
print(lengths)
```

## 10. Nested dictionaries

A value can itself be a dictionary. Chain keys to access nested values; use `get()` when a key might be missing.

```python
students = {
    "S101": {"name": "Aman", "marks": 90},
    "S102": {"name": "Riya", "marks": 95},
}
print(students["S101"]["name"])
print(students["S102"]["marks"])

students["S101"]["marks"] = 92
students["S103"] = {"name": "Kabir", "marks": 78}
print(students)
print(students.get("S999", {}).get("name", "Not found"))
```

## 11. Useful built-in functions

- `len(d)`: number of entries
- `sorted(d)`: sorted list of keys
- `min(d)` / `max(d)`: smallest/largest comparable key
- `sum(d.values())`: sum numeric values
- `any(d)` / `all(d)`: test truthiness of keys
- `list(d)`: list of keys
- `type(d)`: inspect type

Remember: `sum(d)` iterates over keys, not values.

```python
scores = {"Aman": 90, "Riya": 95, "Kabir": 78}
print(len(scores))
print(sorted(scores))
print(list(scores))
print(sum(scores.values()))
print(sum(scores.values()) / len(scores))
print(any(scores), all(scores))
print(type(scores))
```

## 12. Common mistakes

1. Using `d[key]` when a key may be absent; use `get()` or check membership.
2. Thinking `key in d` checks values; it checks keys.
3. Expecting duplicate keys to be retained; keys are unique.
4. Confusing `update()` with copying; `update()` modifies the dictionary.
5. Forgetting that `pop()` needs a key.
6. Forgetting `popitem()` removes the last inserted pair.
7. Using a list or dictionary as a key; mutable unhashable objects cannot be keys.
8. Treating `keys()`, `values()`, and `items()` as ordinary lists; they are views.
9. Forgetting shallow copies share nested mutable values.
10. Using `sum(d)` when you mean `sum(d.values())`.

```python
d = {"a": 10, "b": 20}
print("a" in d)
print(10 in d.values())
print(sum(d.values()))
print(d.get("c", 0))

# Invalid: list keys are unhashable
# invalid = {[1, 2]: "value"}
```
