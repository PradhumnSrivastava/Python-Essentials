## 1. List creation and fundamentals

Lists use square brackets. They preserve order, are mutable, and allow duplicates.

```python
numbers = [10, 20, 30, 20]
names = ["Aman", "Riya", "Kabir"]
mixed = [10, "Python", 3.14, True, None]
empty = []

print(numbers)
print(type(numbers))
print(len(numbers))
print(mixed)
print(list("Python"))
print(list(range(2, 11, 2)))
print("red,green,blue".split(","))
```

## 2. Indexing and slicing

Positive indices start at 0; negative indices count from the end. Slicing syntax is `items[start:stop:step]`; the stop index is excluded. Indexing outside the valid range raises `IndexError`, while slice boundaries can be out of range.

```python
items = ["a", "b", "c", "d", "e", "f"]
print(items[0], items[-1], items[-2])
print(items[1:4])
print(items[:3])
print(items[3:])
print(items[::2])
print(items[::-1])
print(items[-4:-1])
print(items[100:200])
```

## 3. Updating and deleting

Lists are mutable. Assign by index, assign to slices, or delete with `del`. Slice assignment can change the length of a list.

```python
items = [10, 20, 30, 40]
items[1] = 99
print(items)
items[1:3] = [200, 300, 400]
print(items)
del items[0]
print(items)
del items[1:3]
print(items)
```

## 4. Adding elements

- `append(x)`: add `x` as one element.
- `extend(iterable)`: add every element from an iterable.
- `insert(i, x)`: insert at an index.
- `+`: create a concatenated list.
- `+=`: extend the existing list in place.

```python
a = [1, 2]
a.append([3, 4])
print("append:", a)

b = [1, 2]
b.extend([3, 4])
print("extend:", b)

c = [10, 30]
c.insert(1, 20)
print("insert:", c)
print([1, 2] + [3, 4])
c += [40]
print(c)
```

## 5. Removing elements

- `remove(x)`: removes the first matching value; raises `ValueError` if absent.
- `pop()` / `pop(i)`: removes and returns the last element or indexed element.
- `clear()`: empties the list.
- `del items[i]` or `del items[a:b]`: deletes an item or slice.

```python
items = [10, 20, 30, 20, 40]
items.remove(20)
print(items)
removed = items.pop()
print("popped:", removed, items)
removed = items.pop(1)
print("popped index 1:", removed, items)
items.clear()
print("cleared:", items)
```

## 6. Searching and counting

`in`, `not in`, `index(value)`, and `count(value)` are commonly used. `index()` raises `ValueError` if the value is absent.

```python
items = ["red", "blue", "green", "blue"]
print("blue" in items)
print("orange" not in items)
print(items.index("blue"))
print(items.count("blue"))

target = "orange"
if target in items:
    print(items.index(target))
else:
    print("Not found")
```

## 7. Sorting and reversing

- `list.sort()` sorts in place and returns `None`.
- `sorted(list)` returns a new sorted list.
- `list.reverse()` reverses in place.
- `reversed(list)` returns a reverse iterator.
- `key=` customizes sorting; `reverse=True` sorts descending.

```python
numbers = [5, 2, 9, 1, 5]
result = numbers.sort()
print(numbers, result)

numbers = [5, 2, 9, 1, 5]
print("new sorted:", sorted(numbers))
print("descending:", sorted(numbers, reverse=True))
words = ["banana", "fig", "apple", "kiwi"]
print("by length:", sorted(words, key=len))
numbers.reverse()
print("reversed in place:", numbers)
print(list(reversed(numbers)))
```

```python
students = [
    {"name": "Riya", "marks": 88},
    {"name": "Aman", "marks": 95},
    {"name": "Kabir", "marks": 76},
]
print(sorted(students, key=lambda student: student["marks"]))
print(sorted(students, key=lambda student: student["marks"], reverse=True))
```

## 8. Copying, aliasing, shallow copy, and deep copy

`b = a` creates another reference to the same list. `a.copy()` and `a[:]` create shallow copies. A shallow copy does not recursively copy nested mutable objects; `copy.deepcopy()` can be used when a recursive copy is needed.

```python
import copy

a = [1, 2, 3]
b = a
b[0] = 99
print("alias:", a, b)

a = [1, 2, 3]
b = a.copy()
b[0] = 99
print("shallow copy:", a, b)

nested = [[1, 2], [3, 4]]
shallow = nested.copy()
deep = copy.deepcopy(nested)
shallow[0][0] = 100
print("nested after shallow edit:", nested)
deep[1][0] = 300
print("nested after deep edit:", nested)
print("deep:", deep)
```

## 9. List operators and built-in functions

Operators: `+`, `*`, `in`, `not in`, equality and lexicographical comparisons, `+=`, and `del`.

Useful built-ins: `len`, `min`, `max`, `sum`, `sorted`, `reversed`, `any`, `all`, `enumerate`, `zip`, `list`, `map`, `filter`, `iter`, `next`, `type`, and `id`.

`sum()` is for numeric values; `min()` and `max()` require comparable values.

```python
numbers = [4, 1, 9, 2]
print([1, 2] + [3, 4])
print(["x", "y"] * 2)
print(2 in numbers)
print(numbers == [4, 1, 9, 2])
print(len(numbers), min(numbers), max(numbers), sum(numbers))
print(sorted(numbers))
print(list(reversed(numbers)))
print(any([0, "", None, 5]))
print(all([1, "yes", True]))
print(list(enumerate(numbers)))
print(list(zip(["Aman", "Riya"], [80, 95])))
print(list(map(lambda x: x * 2, numbers)))
print(list(filter(lambda x: x % 2 == 0, numbers)))
it = iter(numbers)
print(next(it), next(it))
```

## 10. All 11 standard list methods

1. `append(x)`
2. `extend(iterable)`
3. `insert(i, x)`
4. `remove(x)`
5. `pop([i])`
6. `clear()`
7. `index(x[, start[, end]])`
8. `count(x)`
9. `sort(*, key=None, reverse=False)`
10. `reverse()`
11. `copy()`

Most mutating methods return `None`, rather than the modified list.

```python
a = [1, 2]; a.append(3); print("append", a)
a = [1, 2]; a.extend([3, 4]); print("extend", a)
a = [10, 30]; a.insert(1, 20); print("insert", a)
a = [1, 2, 2, 3]; a.remove(2); print("remove", a)
a = [10, 20, 30]; print("pop", a.pop(), a); print("pop(0)", a.pop(0), a)
a = [1, 2]; a.clear(); print("clear", a)
a = [5, 8, 5, 9]; print("index", a.index(5)); print("index from 1", a.index(5, 1))
print("count", a.count(5))
a = [4, 1, 3, 2]; ret = a.sort(); print("sort", a, "return:", ret)
a.reverse(); print("reverse", a)
b = a.copy(); b.append(100); print("copy", a, b)
```

## 11. Iteration and unpacking

Use direct iteration when you only need values; use `enumerate()` for index and value; use `zip()` to process iterables together. Unpacking assigns elements to multiple variables. Starred unpacking captures remaining elements.

```python
items = ["red", "green", "blue"]
for item in items:
    print(item)

for index, item in enumerate(items):
    print(index, item)

for name, score in zip(["Aman", "Riya"], [80, 95]):
    print(name, score)

x, y = [10, 20]
first, *middle, last = [1, 2, 3, 4, 5]
print(x, y)
print(first, middle, last)
```

## 12. List comprehensions

General form: `[expression for item in iterable if condition]`.

Use these for concise transformations and filtering. A normal loop may be clearer when logic is complex.

```python
numbers = list(range(10))
squares = [n * n for n in numbers]
evens = [n for n in numbers if n % 2 == 0]
labels = ["even" if n % 2 == 0 else "odd" for n in numbers]
print(squares)
print(evens)
print(labels)

matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat = [value for row in matrix for value in row]
doubled = [[value * 2 for value in row] for row in matrix]
print(flat)
print(doubled)
```

## 13. Nested lists and 2D data

Nested lists can represent matrices or tables. When creating a grid, make independent rows. Avoid `[[0] * columns] * rows` if each row must be independently mutable.

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(matrix[0])
print(matrix[1][2])
for row in matrix:
    print(row)
print("first column:", [row[0] for row in matrix])

bad_grid = [[0] * 3] * 2
bad_grid[0][0] = 99
print("shared rows:", bad_grid)

grid = [[0 for _ in range(3)] for _ in range(2)]
grid[0][0] = 99
print("independent rows:", grid)
```

## 14. Useful list patterns

```python
# Remove duplicates while preserving order
values = [3, 1, 3, 2, 1, 4]
unique = []
for value in values:
    if value not in unique:
        unique.append(value)
print("unique:", unique)

# Frequency counting
values = ["red", "blue", "red", "green", "blue", "red"]
frequency = {}
for value in values:
    frequency[value] = frequency.get(value, 0) + 1
print("frequency:", frequency)

# Separate even and odd numbers
numbers = [1, 2, 3, 4, 5, 6]
print([n for n in numbers if n % 2 == 0])
print([n for n in numbers if n % 2 != 0])

# Second-largest distinct value
numbers = [10, 5, 10, 8, 3]
distinct_sorted = sorted(set(numbers))
print(distinct_sorted[-2] if len(distinct_sorted) >= 2 else None)
```

## 15. Common mistakes

1. `append([3, 4])` adds one nested list; `extend([3, 4])` adds two items.
2. `sort()` mutates and returns `None`; use `sorted()` when you need a new list.
3. `b = a` is aliasing, not copying.
4. Shallow copies still share nested mutable objects.
5. `remove()` raises `ValueError` if the value is absent.
6. `pop()` on an empty list raises `IndexError`.
7. Mutating a list while iterating can skip elements; often build a new list instead.
8. Repeating a nested list can create shared references.
9. Sorting incomparable mixed types can raise `TypeError`.
10. Front insertion and front popping are generally O(n).

```python
a = [1, 2]
a.append([3, 4])
print("append:", a)

b = [1, 2]
b.extend([3, 4])
print("extend:", b)

c = [3, 1, 2]
result = c.sort()
print("sorted in place:", c, "return:", result)

values = [1, 2, 3, 4, 5, 6]
values = [value for value in values if value % 2 != 0]
print("filtered safely:", values)
```

## 16. Time complexity overview

Approximate typical costs:

| Operation | Complexity |
|---|---:|
| Index access/update | O(1) |
| `append()` | Amortized O(1) |
| `pop()` at end | O(1) |
| `insert(0, x)` | O(n) |
| `pop(0)` | O(n) |
| Membership, `index()`, `count()`, `remove()` | O(n) |
| Copying a list | O(n) |
| Slice of k elements | O(k) |
| Sorting | O(n log n) typical |
| `reverse()` | O(n) |
