# Python Dictionary: Important Operations

**Beginner-friendly notes with simple code and comments**

A dictionary stores information in **key-value pairs**.

Example: a student's name is a key, and their marks are the value.

Run the code cells one by one and read the comments.

## 1. Create a dictionary

```python
# Create a dictionary with student details
student = {
    "name": "Aman",
    "age": 18,
    "marks": 85
}

# Display the complete dictionary
print(student)
```

## 2. Access values

```python
student = {"name": "Aman", "age": 18, "marks": 85}

# Access a value using its key
print(student["name"])
print(student["marks"])

# get() safely returns a value for the given key
print(student.get("age"))

# If the key is missing, get() returns None by default
print(student.get("city"))
```

## 3. Add a new item

```python
student = {"name": "Aman", "age": 18}

# Add a new key-value pair
student["city"] = "Delhi"

print(student)
```

## 4. Update an existing value

```python
student = {"name": "Aman", "marks": 85}

# Change the value of the marks key
student["marks"] = 92

print(student)
```

## 5. Add or update with `update()`

```python
student = {"name": "Aman", "marks": 85}

# update() can add a new key or change an existing value
student.update({"marks": 90, "city": "Delhi"})

print(student)
```

## 6. Remove an item with `pop()`

```python
student = {"name": "Aman", "age": 18, "marks": 85}

# Remove the item with the key "age"
removed_age = student.pop("age")

print("Removed age:", removed_age)
print("Updated dictionary:", student)
```

## 7. Remove an item with `del`

```python
student = {"name": "Aman", "age": 18, "marks": 85}

# Delete the item whose key is "marks"
del student["marks"]

print(student)
```

## 8. Remove the last inserted item with `popitem()`

```python
student = {"name": "Aman", "age": 18, "marks": 85}

# popitem() removes and returns the last inserted key-value pair
last_item = student.popitem()

print("Removed item:", last_item)
print("Updated dictionary:", student)
```

## 9. Get all keys, values, and items

```python
student = {"name": "Aman", "age": 18, "marks": 85}

# keys() gives all keys
print(student.keys())

# values() gives all values
print(student.values())

# items() gives key-value pairs
print(student.items())
```

## 10. Loop through a dictionary

```python
student = {"name": "Aman", "age": 18, "marks": 85}

# Loop through keys
for key in student:
    print(key)

print("---")

# Loop through keys and values together
for key, value in student.items():
    print(key, ":", value)
```

## 11. Check whether a key exists

```python
student = {"name": "Aman", "age": 18}

# Use 'in' to check whether a key is present
if "name" in student:
    print("Name is available")

# Check for a key that does not exist
if "marks" not in student:
    print("Marks are not available")
```

## 12. Find the number of items

```python
student = {"name": "Aman", "age": 18, "marks": 85}

# len() returns the number of key-value pairs
print(len(student))
```

## 13. Make a copy

```python
student = {"name": "Aman", "marks": 85}

# copy() creates a separate, shallow copy of the dictionary
student_copy = student.copy()

# Change the copy
student_copy["marks"] = 95

print("Original:", student)
print("Copy:", student_copy)
```

## 14. Remove all items with `clear()`

```python
student = {"name": "Aman", "age": 18}

# clear() removes every item from the dictionary
student.clear()

print(student)
```

## 15. Use `setdefault()`

```python
student = {"name": "Aman", "age": 18}

# If "city" is missing, setdefault() adds it with the value "Delhi"
student.setdefault("city", "Delhi")

# "name" already exists, so its value does not change
student.setdefault("name", "Ravi")

print(student)
```

## 16. Simple dictionary example: student marks

```python
# Store marks for three subjects
marks = {
    "English": 78,
    "Maths": 88,
    "Science": 91
}

# Print each subject and its marks
for subject, score in marks.items():
    print(subject, ":", score)

# Find the total marks
total = sum(marks.values())
print("Total marks:", total)

# Find the average marks
average = total / len(marks)
print("Average marks:", average)
```

## Important functions and methods recap

| Operation | What it does |
|---|---|
| `dict()` | Creates a dictionary |
| `get(key)` | Gets a value safely |
| `keys()` | Returns the keys |
| `values()` | Returns the values |
| `items()` | Returns key-value pairs |
| `update()` | Adds or updates items |
| `pop(key)` | Removes an item by key |
| `popitem()` | Removes the last inserted item |
| `setdefault(key, value)` | Adds a default value if the key is missing |
| `copy()` | Makes a shallow copy |
| `clear()` | Removes all items |
| `len(dictionary)` | Counts the key-value pairs |

**Remember:** Dictionary keys must be unique. If you assign a value to a key that already exists, its old value is replaced.
